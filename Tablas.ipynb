```python
import pandas as pd
```

## LOGIN TABLA MATERIA - DETALLE_MATERIA

```python
# Rutas relativas a los archivos dentro de la carpeta "db"
ruta_archivo1 = "db/tablas_procesadas/Materia.xlsx"  
ruta_archivo2 = "db/tablas_procesadas/Detalle_materia.xlsx"  
```

```python
df_materias = pd.read_excel(ruta_archivo1, engine='openpyxl')
df_detalle_materia = pd.read_excel(ruta_archivo2, engine='openpyxl')
```

```python
# Realizar el merge (login) usando "materia_id" como clave
df_merged = pd.merge(
    df_materias,         # Tabla izquierda (materias)
    df_detalle_materia, # Detalle solo de las regulares
    on="id_materia",     # Columna común
    how="inner"          # Tipo de join (inner para datos que coincidan en ambas tablas)
)
```

## LOGIN TABLA PROFESOR_MATERIA

```python
ruta_archivo4 = "db/tablas_procesadas/Profesor_materia.xlsx"
df_profesor_materia = pd.read_excel(ruta_archivo4, engine='openpyxl')
```

```python
ruta_archivo5 = "db/tablas_procesadas/Profesor.xlsx"
df_profesor = pd.read_excel(ruta_archivo5, engine='openpyxl')
```

```python
df_merged = pd.merge(
    df_merged,                  # DataFrame con toda la información previa
    df_profesor,        # Nueva tabla de profesor asociado a su materia
    on="id_profesor",             # Columna común
    how="left"                 # Mantener todos los registros de df_final
)
```

```python
df_merged
```

```python
# Eliminar columnas no necesarias
df_merged = df_merged.drop(columns=["id_materia", "tipo_materia", "curso", "id_curso", "detalle", "nivel", "inscritos","id_profesor", "seccion", "categoria", "nombre", "grupos"])  

# Verificar las columnas restantes
print("Columnas finales:", df_merged.columns.tolist())
```

```python
columnas = [
    'nombre_materia', 
    'pos_prof', 
    'pos_curso', 
    'laboratorio', 
    'horas_semanales_tipicas',  
    'clase_seguida', 
    'aula_req'
]

# Reordenar el DataFrame
df_merged = df_merged[columnas]
```

```python
df_merged
```

```python
df_merged.insert(0, 'id_curso', ['C' + str(i + 1) for i in range(len(df_merged))])

df_merged 
```

```python
# Genera el archivo courses.txt
with open("courses.txt", "w") as file:
    for _, row in df_merged.iterrows():
        course_str = f'new Course("{row["id_curso"]}", "{row["nombre_materia"]}", professors[{row["pos_prof"]}]'
        
        # Agrega `pos_curso` si no es nulo
        if pd.notna(row["pos_curso"]):
            course_str += f', studentGroups[{int(row["pos_curso"])}]'
        else:
            course_str += ', null'  # O puedes omitirlo si prefieres
        
        course_str += f', {str(row["laboratorio"]).lower()}, {row["horas_semanales_tipicas"]}, {str(row["clase_seguida"]).lower()}'
        
        # Agrega el valor de `aula_req` si no es nulo
        if pd.notna(row["aula_req"]):
            course_str += f', "{row["aula_req"]}"'
        
        course_str += '),\n'
        
        # Escribe la línea en el archivo
        file.write(course_str)

print("Archivo courses.txt generado con éxito.")
```

```python
ruta = "db/tablas_procesadas/Curso.xlsx"
data_curso = pd.read_excel(ruta, engine='openpyxl')
```

```python
with open("student_groups.txt", "w") as file:
    for _, row in data_curso.iterrows():
        group_str = f'new StudentGroup("{row["id_curso"]}", "{row["curso"]}", {row["inscritos"]}),\n'
        file.write(group_str)

print("Archivo student_groups.txt generado con éxito.")
```

```python
ruta2 = "db/tablas_procesadas/Profesor.xlsx"
data_profesor = pd.read_excel(ruta2, engine='openpyxl')
```

```python
with open("professors.txt", "w") as file:
    for _, row in data_profesor.iterrows():
        # Construir la declaración de Professor
        professor_str = f'new Professor("{row["id_profesor"]}", "{row["nombre"]}"),\n'
        # Escribir en el archivo
        file.write(professor_str)

print("Archivo professors.txt generado con éxito.")
```

```python
ruta3 = "db/tablas_procesadas/Sala.xlsx"
data_sala = pd.read_excel(ruta3, engine='openpyxl')
```

```python
with open("rooms.txt", "w") as file:
    for _, row in data_sala.iterrows():
        # Construir la declaración de Room
        room_str = f'new Room("{row["sala_id"]}", {row["capacidad"]}, {str(row["laboratorio"]).lower()}),\n'
        # Escribir en el archivo
        file.write(room_str)

print("Archivo rooms.txt generado con éxito.")
```
